```python
def eligibility(gender, age):
    if (gender.lower() == 'male' and age >= 21) or (gender.lower() == 'female' and age >= 18):
        print("ELIGIBLE")
    else:
        print("NOT ELIGIBLE")

# Usage
eligibility('Male', 20)
```

```python
def triangle_area(height, breadth):
    area = (height * breadth) / 2
    print("Area of Triangle: ", area)

def triangle_perimeter(h1, h2, breadth):
    perimeter = h1 + h2 + breadth
    print("Perimeter of Triangle: ", perimeter)

# Usage
triangle_area(32, 34)
triangle_perimeter(2, 4, 4)
```
